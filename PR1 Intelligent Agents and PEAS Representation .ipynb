```python
peas = {
    "Performance Measure": "Comfortable temperature and energy saving",
    "Environment": "Room",
    "Actuators": ["AC ON", "AC OFF"],
    "Sensors": ["Temperature Sensor", "Occupancy Sensor"]
}

for component, details in peas.items():
    print(component, ":", details)
```

```python
def smart_home_agent(temperature, occupied, ac_status):

    if not occupied:
        return "AC OFF"

    elif temperature > 28:
        return "AC ON"

    elif temperature < 24:
        return "AC OFF"

    else:
        return ac_status
```

```python
temperature = 32
occupied = True
ac_status = "AC OFF"

action = smart_home_agent(
    temperature, occupied, ac_status
)

print("Temperature:", temperature, "°C")
print("Room Occupied:", occupied)
print("Agent Decision:", action)
```

```python
test_cases = [
    (32, True, "AC OFF"),
    (20, True, "AC ON"),
    (26, True, "AC ON"),
    (32, False, "AC ON")
]

for temp, occupied, status in test_cases:
    result = smart_home_agent(temp, occupied, status)
    print(f"Temperature: {temp}°C | Occupied: {occupied} | Action: {result}")
```
