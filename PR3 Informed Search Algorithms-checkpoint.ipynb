```python
graph = {
    'S': {'A': 1, 'B': 4},
    'A': {'C': 2, 'D': 4},
    'B': {'G': 5},
    'C': {'G': 10},
    'D': {'G': 3},
    'G': {}
}
```

```python
heuristic = {
    'S': 7,
    'A': 6,
    'B': 4,
    'C': 8,
    'D': 2,
    'G': 0
}
```

```python
import heapq

def greedy_best_first_search(graph, heuristic, start, goal):
    pq = [(heuristic[start], start)]
    parent = {start: None}
    visited = set()

    while pq:
        _, current = heapq.heappop(pq)

        if current in visited:
            continue

        visited.add(current)

        if current == goal:
            break

        for neighbor in graph[current]:
            if neighbor not in visited and neighbor not in parent:
                parent[neighbor] = current
                heapq.heappush(
                    pq, (heuristic[neighbor], neighbor)
                )

    if goal not in visited:
        return None, 0, len(visited)

    path = []
    node = goal

    while node is not None:
        path.append(node)
        node = parent[node]

    path.reverse()

    cost = sum(
        graph[path[i]][path[i + 1]]
        for i in range(len(path) - 1)
    )

    return path, cost, len(visited)
```

```python
def a_star_search(graph, heuristic, start, goal):
    pq = [(heuristic[start], 0, start)]
    parent = {start: None}
    best_g = {start: 0}
    visited = set()

    while pq:
        f, g, current = heapq.heappop(pq)

        if g != best_g[current]:
            continue

        if current in visited:
            continue

        visited.add(current)

        if current == goal:
            break

        for neighbor, cost in graph[current].items():
            new_g = g + cost

            if new_g < best_g.get(neighbor, float('inf')):
                best_g[neighbor] = new_g
                parent[neighbor] = current

                new_f = new_g + heuristic[neighbor]

                heapq.heappush(
                    pq, (new_f, new_g, neighbor)
                )

    if goal not in visited:
        return None, 0, len(visited)

    path = []
    node = goal

    while node is not None:
        path.append(node)
        node = parent[node]

    path.reverse()

    return path, best_g[goal], len(visited)
```

```python
start = 'S'
goal = 'G'

gbfs_path, gbfs_cost, gbfs_nodes = greedy_best_first_search(
    graph, heuristic, start, goal
)

astar_path, astar_cost, astar_nodes = a_star_search(
    graph, heuristic, start, goal
)

print("Greedy Best-First Search")
print("Path:", gbfs_path)
print("Cost:", gbfs_cost)
print("Nodes explored:", gbfs_nodes)

print("\nA* Search")
print("Path:", astar_path)
print("Cost:", astar_cost)
print("Nodes explored:", astar_nodes)
```
