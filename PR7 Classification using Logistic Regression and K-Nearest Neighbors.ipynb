```python

# 1. Import libraries

import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)


# 2. Create Dataset

data = {
    'Study_Hours': [1, 2, 3, 4, 5, 6, 7, 8, 2, 6, 3, 7],
    'Attendance': [40, 45, 50, 60, 65, 70, 80, 90, 35, 75, 55, 85],
    'Result': ['Fail', 'Fail', 'Fail', 'Pass', 'Pass', 'Pass',
               'Pass', 'Pass', 'Fail', 'Pass', 'Fail', 'Pass']
}

df = pd.DataFrame(data)

print("Original Dataset:")
print(df)


# 3. Separate Input and Output

X = df[['Study_Hours', 'Attendance']]
y = df['Result']


# 4. Split Training and Testing Data

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.25,
    random_state=42,
    stratify=y
)


# 5. Data Preprocessing (Feature Scaling)

scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)


# 6. Create Models

log_model = LogisticRegression()

knn_model = KNeighborsClassifier(n_neighbors=3)


# 7. Train Models

log_model.fit(X_train, y_train)

knn_model.fit(X_train, y_train)


# 8. Make Predictions

log_pred = log_model.predict(X_test)

knn_pred = knn_model.predict(X_test)


# 9. Evaluate Models

def evaluate_model(name, actual, predicted):

    print("\n", name)

    print("Accuracy:",
          accuracy_score(actual, predicted))

    print("Precision:",
          precision_score(actual, predicted, pos_label='Pass'))

    print("Recall:",
          recall_score(actual, predicted, pos_label='Pass'))

    print("F1-Score:",
          f1_score(actual, predicted, pos_label='Pass'))

    print("Confusion Matrix:")

    print(confusion_matrix(
        actual, predicted,
        labels=['Fail', 'Pass']
    ))


# Evaluate Logistic Regression

evaluate_model("Logistic Regression", y_test, log_pred)


# Evaluate KNN

evaluate_model("KNN", y_test, knn_pred)


# 10. Compare Results

print("\nModel Comparison")

print("Logistic Regression Predictions:", list(log_pred))

print("KNN Predictions:", list(knn_pred))
```
